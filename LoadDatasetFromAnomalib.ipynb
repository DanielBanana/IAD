# This file is for downloading datasets with the help of Anomalib

```python
from anomalib.data import (BMAD,
    MPDD,
    VAD,
    BTech,
    Datumaro,
    Folder,
    ImageDataFormat,
    Kaputt,
    Kolektor,
    MVTecAD,
    MVTecAD2,
    MVTecLOCO,
    RealIAD,
    Tabular,
    Visa
    )

datasetList = [
    # BMAD,
    # MPDD,
    # BTech,
    # Kaputt,
    Kolektor,
    MVTecAD,
    # MVTecAD2,
    # MVTecLOCO,
    RealIAD,
    # Visa
]

for cls in datasetList:
    try:
        datamodule = cls(
        root=f"./datasets/{cls.__name__}",
        train_batch_size=32,
        eval_batch_size=32,
        num_workers=8
        )
        datamodule.prepare_data()
        print(f"Data for {cls.__name__} prepared successfully.")
    except Exception as e:
        print(f"Error preparing data for {cls.__name__}: {e}")
```
